```python
import pandas as pd
import numpy as np

df = pd.read_csv("/content/Retail and wherehouse Sale.csv")

# Remove duplicates
df = df.drop_duplicates()

# finding null values

null_columns = df.isnull().sum()
null_columns = null_columns[null_columns > 0]

print(null_columns)

# Fill missing values

df['SUPPLIER'] = df['SUPPLIER'].fillna(0)
df['RETAIL SALES'] =df['RETAIL SALES'].fillna(0)
null_counts_after_fill = df.isnull().sum() # Assign to a new variable
print(null_counts_after_fill)


df["Month Name"] = pd.to_datetime(
    df["MONTH"], format="%m"
).dt.strftime("%b")

print(df[["MONTH", "Month Name"]].head())

df["Total_Sale"] = (df["RETAIL SALES"] + df['WAREHOUSE SALES'])

df["Supplier Sales"] = (
    df.groupby("SUPPLIER")["RETAIL SALES"]
      .transform("sum")
)


Q1 = df["RETAIL SALES"].quantile(0.25)
Q3 = df["RETAIL SALES"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df["Outlier"] = (
    (df["RETAIL SALES"] < lower) |
    (df["RETAIL SALES"] > upper)
)

df.to_csv("Retail_Sales_Cleaned.csv", index=False)
```
